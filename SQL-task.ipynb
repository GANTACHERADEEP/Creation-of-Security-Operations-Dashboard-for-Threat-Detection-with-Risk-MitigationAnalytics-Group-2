Write SQL queries to demonstrate Aggregate Functions. Create an Employees table, insert sample records, and write queries to calculate the total number of employees, average salary, highest salary, lowest salary, and total salary.

```python
import sqlite3

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# 1. Create the Employees table
cursor.execute('''
CREATE TABLE Employees (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    department TEXT,
    salary REAL NOT NULL
);
''')
print("Table 'Employees' created successfully.\n")

# 2. Insert sample records
employees_data = [
    ('Alice', 'HR', 60000),
    ('Bob', 'Engineering', 90000),
    ('Charlie', 'HR', 65000),
    ('David', 'Sales', 75000),
    ('Eve', 'Engineering', 100000),
    ('Frank', 'Sales', 80000)
]

cursor.executemany('INSERT INTO Employees (name, department, salary) VALUES (?, ?, ?)', employees_data)
conn.commit()
print("Sample records inserted successfully.\n")

# 3. Write queries to demonstrate Aggregate Functions

# Total number of employees
cursor.execute('SELECT COUNT(*) FROM Employees;')
total_employees = cursor.fetchone()[0]
print(f"Total Number of Employees: {total_employees}")

# Average salary
cursor.execute('SELECT AVG(salary) FROM Employees;')
average_salary = cursor.fetchone()[0]
print(f"Average Salary: {average_salary:.2f}")

# Highest salary
cursor.execute('SELECT MAX(salary) FROM Employees;')
highest_salary = cursor.fetchone()[0]
print(f"Highest Salary: {highest_salary:.2f}")

# Lowest salary
cursor.execute('SELECT MIN(salary) FROM Employees;')
lowest_salary = cursor.fetchone()[0]
print(f"Lowest Salary: {lowest_salary:.2f}")

# Total salary
cursor.execute('SELECT SUM(salary) FROM Employees;')
total_salary = cursor.fetchone()[0]
print(f"Total Salary: {total_salary:.2f}")

# Close the connection
conn.close()
```

Write SQL queries to demonstrate GROUP BY and HAVING clauses. Create an Employees table with sample data and display the average salary of each department. Show only departments whose average salary is greater than 50,000.


```python
import sqlite3

# Connect to an in-memory SQLite database for this demonstration
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# 1. Create the Employees table (if not already exists or for a fresh start)
cursor.execute('''
CREATE TABLE Employees (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    department TEXT,
    salary REAL NOT NULL
);
''')
print("Table 'Employees' created successfully.\n")

# 2. Insert sample records
employees_data = [
    ('Alice', 'HR', 60000),
    ('Bob', 'Engineering', 90000),
    ('Charlie', 'HR', 65000),
    ('David', 'Sales', 75000),
    ('Eve', 'Engineering', 100000),
    ('Frank', 'Sales', 80000),
    ('Grace', 'HR', 45000), # Adding an employee to HR to potentially lower average
    ('Heidi', 'Sales', 50000) # Adding an employee to Sales
]

cursor.executemany('INSERT INTO Employees (name, department, salary) VALUES (?, ?, ?)', employees_data)
conn.commit()
print("Sample records inserted successfully.\n")

# 3. Demonstrate GROUP BY and HAVING clauses
# Calculate average salary per department and filter those > 50000
cursor.execute('''
SELECT department, AVG(salary) AS avg_dept_salary
FROM Employees
GROUP BY department
HAVING AVG(salary) > 50000;
''')

print("Departments with average salary greater than 50,000:\n")
print("%-15s %s" % ("Department", "Average Salary"))
print("------------------------------")
for row in cursor.fetchall():
    print("%-15s %.2f" % (row[0], row[1]))

# Close the connection
conn.close()
```

Write SQL queries to demonstrate INNER JOIN. Create Customers and Orders tables with sample data and display customer names along with their order IDs and order amounts.

```python
import sqlite3

# Connect to an in-memory SQLite database for this demonstration
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# 1. Create the Customers table
cursor.execute('''
CREATE TABLE Customers (
    customer_id INTEGER PRIMARY KEY,
    customer_name TEXT NOT NULL
);
''')
print("Table 'Customers' created successfully.\n")

# 2. Insert sample customers
customers_data = [
    (1, 'Alice Smith'),
    (2, 'Bob Johnson'),
    (3, 'Charlie Brown')
]
cursor.executemany('INSERT INTO Customers (customer_id, customer_name) VALUES (?, ?)', customers_data)
conn.commit()
print("Sample customers inserted successfully.\n")

# 3. Create the Orders table with a foreign key to Customers
cursor.execute('''
CREATE TABLE Orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER NOT NULL,
    order_amount REAL NOT NULL,
    FOREIGN KEY (customer_id) REFERENCES Customers(customer_id)
);
''')
print("Table 'Orders' created successfully.\n")

# 4. Insert sample orders
orders_data = [
    (101, 1, 150.00),
    (102, 2, 200.50),
    (103, 1, 75.25),
    (104, 3, 300.00),
    (105, 2, 50.00)
]
cursor.executemany('INSERT INTO Orders (order_id, customer_id, order_amount) VALUES (?, ?, ?)', orders_data)
conn.commit()
print("Sample orders inserted successfully.\n")

# 5. Demonstrate INNER JOIN
# Display customer names along with their order IDs and order amounts
cursor.execute('''
SELECT
    C.customer_name,
    O.order_id,
    O.order_amount
FROM
    Customers AS C
INNER JOIN
    Orders AS O ON C.customer_id = O.customer_id;
''')

print("Customer Names with their Order IDs and Amounts:\n")
print("%-20s %-10s %s" % ("Customer Name", "Order ID", "Order Amount"))
print("---------------------------------------------------")
for row in cursor.fetchall():
    print("%-20s %-10d %.2f" % (row[0], row[1], row[2]))

# Close the connection
conn.close()
```

Write SQL queries to demonstrate LEFT JOIN, RIGHT JOIN, and FULL OUTER JOIN using Employees and Departments tables with sample data.

```python
import sqlite3

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# 1. Create Departments table
cursor.execute('''
CREATE TABLE Departments (
    dept_id INTEGER PRIMARY KEY,
    dept_name TEXT NOT NULL
);
''')
print("Table 'Departments' created successfully.\n")

# 2. Insert sample departments
departments_data = [
    (1, 'HR'),
    (2, 'Engineering'),
    (3, 'Sales'),
    (4, 'Marketing') # Marketing has no employees yet
]
cursor.executemany('INSERT INTO Departments (dept_id, dept_name) VALUES (?, ?)', departments_data)
conn.commit()
print("Sample departments inserted successfully.\n")

# 3. Create Employees table
cursor.execute('''
CREATE TABLE Employees (
    emp_id INTEGER PRIMARY KEY,
    emp_name TEXT NOT NULL,
    dept_id INTEGER,
    FOREIGN KEY (dept_id) REFERENCES Departments(dept_id)
);
''')
print("Table 'Employees' created successfully.\n")

# 4. Insert sample employees
employees_data = [
    (101, 'Alice', 1), # HR
    (102, 'Bob', 2),   # Engineering
    (103, 'Charlie', 1), # HR
    (104, 'David', 3),   # Sales
    (105, 'Eve', None),  # No department assigned yet - Corrected from NONe to None
    (106, 'Frank', 2) # Engineering
]
cursor.executemany('INSERT INTO Employees (emp_id, emp_name, dept_id) VALUES (?, ?, ?)', employees_data)
conn.commit()
print("Sample employees inserted successfully.\n")

# --- Demonstrate Joins ---

# Helper function to print results
def print_results(title, query_result):
    print(f"\n--- {title} ---")
    # Assuming a consistent number of columns for demonstration
    if query_result:
        # Print header based on the first row's columns
        if len(query_result[0]) == 4:
            print(f"%-10s %-15s %-10s %s" % ("Emp ID", "Emp Name", "Dept ID", "Dept Name"))
            print("-------------------------------------------")
            for row in query_result:
                print(f"%-10s %-15s %-10s %s" % (str(row[0] or 'NULL'), str(row[1] or 'NULL'), str(row[2] or 'NULL'), str(row[3] or 'NULL')))
        elif len(query_result[0]) == 3:
            print(f"%-15s %-10s %s" % ("Emp Name", "Dept ID", "Dept Name"))
            print("---------------------------------")
            for row in query_result:
                print(f"%-15s %-10s %s" % (str(row[0] or 'NULL'), str(row[1] or 'NULL'), str(row[2] or 'NULL')))
    else:
        print("No results found.")


# LEFT JOIN: Returns all rows from the left table, and the matched rows from the right table.
# If there's no match, NULL is used for columns from the right table.
cursor.execute('''
SELECT E.emp_id, E.emp_name, D.dept_id, D.dept_name
FROM Employees AS E
LEFT JOIN Departments AS D ON E.dept_id = D.dept_id;
''')
print_results("LEFT JOIN (All Employees and their Departments)", cursor.fetchall())

# Simulate RIGHT JOIN: Returns all rows from the right table, and the matched rows from the left table.
# If there's no match, NULL is used for columns from the left table.
# In SQLite, we achieve this by swapping tables and using a LEFT JOIN.
cursor.execute('''
SELECT E.emp_id, E.emp_name, D.dept_id, D.dept_name
FROM Departments AS D
LEFT JOIN Employees AS E ON D.dept_id = E.dept_id;
''')
# For better readability for RIGHT JOIN, we might want to reorder columns or handle NULLs in Employees data
# Here, just printing the raw output as it comes from the reversed LEFT JOIN
print_results("RIGHT JOIN (Simulated - All Departments and their Employees)", cursor.fetchall())

# Simulate FULL OUTER JOIN: Returns all rows when there is a match in either of the tables.
# It essentially combines the results of a LEFT JOIN and a RIGHT JOIN.
# In SQLite, we achieve this using UNION ALL.
cursor.execute('''
SELECT E.emp_id, E.emp_name, D.dept_id, D.dept_name
FROM Employees AS E
LEFT JOIN Departments AS D ON E.dept_id = D.dept_id
UNION ALL
SELECT E.emp_id, E.emp_name, D.dept_id, D.dept_name
FROM Departments AS D
LEFT JOIN Employees AS E ON D.dept_id = E.dept_id
WHERE E.emp_id IS NULL;
''')
print_results("FULL OUTER JOIN (Simulated - All Employees and Departments)", cursor.fetchall())

# Close the connection
conn.close()
```

## CRUD Operations (Create, Read, Update, Delete)

```python
import sqlite3

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# --- CREATE (Table and initial data) ---
# 1. Create the Students table
cursor.execute('''
CREATE TABLE Students (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    marks REAL
);
''')
print("Table 'Students' created successfully.\n")

# 2. Insert sample records
students_data = [
    ('Alice', 18, 85.5),
    ('Bob', 19, 72.0),
    ('Charlie', 18, 91.0),
    ('David', 20, 68.5)
]
cursor.executemany('INSERT INTO Students (name, age, marks) VALUES (?, ?, ?)', students_data)
conn.commit()
print("Sample students inserted successfully.\n")

# --- READ (Display initial table) ---
print("--- Initial Students Table ---")
cursor.execute('SELECT * FROM Students;')
print("%-5s %-10s %-5s %s" % ("ID", "Name", "Age", "Marks"))
print("--------------------------------")
for row in cursor.fetchall():
    print("%-5d %-10s %-5d %.2f" % (row[0], row[1], row[2], row[3]))

# --- UPDATE ---
# Update Charlie's marks
cursor.execute("UPDATE Students SET marks = 95.0 WHERE name = 'Charlie';")
conn.commit()
print("\nUpdated Charlie's marks to 95.0")

# --- DELETE ---
# Delete David's record
cursor.execute("DELETE FROM Students WHERE name = 'David';")
conn.commit()
print("Deleted David's record")

# --- READ (Display final table after operations) ---
print("\n--- Final Students Table ---")
cursor.execute('SELECT * FROM Students;')
print("%-5s %-10s %-5s %s" % ("ID", "Name", "Age", "Marks"))
print("--------------------------------")
for row in cursor.fetchall():
    print("%-5d %-10s %-5d %.2f" % (row[0], row[1], row[2], row[3]))

# Close the connection
conn.close()
```

## Add a New Column to the Students Table using the Alter Method


```python
import sqlite3

# Connect to an in-memory SQLite database (re-creating the table as previous connection was closed)
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Re-create the Students table with initial data
cursor.execute('''
CREATE TABLE Students (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    marks REAL
);
''')

students_data = [
    ('Alice', 18, 85.5),
    ('Bob', 19, 72.0),
    ('Charlie', 18, 91.0),
    ('David', 20, 68.5)
]
cursor.executemany('INSERT INTO Students (name, age, marks) VALUES (?, ?, ?)', students_data)
conn.commit()
print("Students table re-created and populated.\n")

# --- ALTER TABLE: Add a new column ---
# Add a 'grade' column to the Students table
cursor.execute('ALTER TABLE Students ADD COLUMN grade TEXT;')
conn.commit()
print("Column 'grade' added to the Students table.\n")

# Verify the new schema
print("--- Updated Table Schema ---")
cursor.execute("PRAGMA table_info(Students);")
for column in cursor.fetchall():
    print(column)
print("\n")

# Update some records to populate the new 'grade' column
cursor.execute("UPDATE Students SET grade = 'A' WHERE name = 'Alice';")
cursor.execute("UPDATE Students SET grade = 'B' WHERE name = 'Bob';")
cursor.execute("UPDATE Students SET grade = 'A' WHERE name = 'Charlie';")
cursor.execute("UPDATE Students SET grade = 'C' WHERE name = 'David';")
conn.commit()
print("Grades updated for students.\n")

# --- Display the final table contents with the new column ---
print("--- Students Table with new 'grade' column ---")
cursor.execute('SELECT * FROM Students;')
print("%-5s %-10s %-5s %-7s %s" % ("ID", "Name", "Age", "Marks", "Grade"))
print("--------------------------------------------------")
for row in cursor.fetchall():
    print("%-5d %-10s %-5d %-7.2f %s" % (row[0], row[1], row[2], row[3], row[4] if row[4] else 'NULL'))

# Close the connection
conn.close()
```
